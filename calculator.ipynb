## Calculator

#### Takes first number

```python
num_1 = eval(input("Enter the first number: "))
```

#### Takes an operator

```python
op = input("Enter any operator - (+,-,*,/)")
```

#### Takes the second number

```python
num_2 = eval(input("Enter the second number: "))
```

##### Evaluate

```python
def evaluate(op):
    match op:
        case "+":
            return f"{num_1} {op} {num_2} = {num_1-num_2}"
        case "-" :
            return f"{num_1} {op} {num_2} = {num_1-num_2}"
        case "-" :
            return f"{num_1} {op} {num_2} = {num_1-num_2}"
        case "*" :
            return f"{num_1} {op} {num_2} = {num_1-num_2}"
        case "/" :
            return f"{num_1} {op} {num_2} = {num_1-num_2}"
        case _:
            return "Invalid operator"

print(f"""
        ===================
        |   Calculator    |
        ===================
        |      M-549  
                           |
        |  {evaluate(op)}  |
        |                 |
        |                 |
        |                 |
        |=================|

        
    """)
```
